```python

# Case Study 1B: Google Calendar Application
# Algorithm: Greedy Activity Selection

def select_meetings(meetings):
    # Sort meetings by ending time
    meetings.sort(key=lambda x: x[2])

    selected = []
    last_end_time = -1

    for meeting in meetings:
        meeting_name, start, end = meeting

        if start >= last_end_time:
            selected.append(meeting)
            last_end_time = end

    return selected


# User Input
meetings = []

n = int(input("Enter number of meetings: "))

for i in range(n):
    print(f"\nEnter details of Meeting {i + 1}")

    name = input("Meeting name: ")
    start = int(input("Start time (24-hour format): "))
    end = int(input("End time (24-hour format): "))

    if start < 0 or end > 24 or start >= end:
        print("Invalid time! Please enter a valid meeting time.")
        continue

    meetings.append((name, start, end))


# Apply Greedy Algorithm
selected_meetings = select_meetings(meetings)

# Display Output
print("\n--- Selected Meetings for Google Calendar ---")

if selected_meetings:
    for name, start, end in selected_meetings:
        print(f"{name}: {start}:00 to {end}:00")
else:
    print("No valid meetings selected.")

print("\nTotal meetings selected:", len(selected_meetings))
```
